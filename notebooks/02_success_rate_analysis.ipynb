# 02 — Success Rate Analysis

Success rates by site type, failure mode breakdown.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/processed/metrics.csv')
df['success'] = df['success'].astype(bool)

# Overall success rate
print('Overall success rate:', df['success'].mean())

# By page type
df.groupby('page_type')['success'].mean().plot(kind='bar', title='Success Rate by Page Type')
plt.ylabel('Success Rate')
plt.tight_layout()
plt.show()
```

```python
# Failure mode distribution
failures = df[~df['success']]
failures['failure_mode'].value_counts().plot(kind='bar', title='Failure Mode Distribution')
plt.tight_layout()
plt.show()
```
