# 03 — Token & Cost Analysis

Token usage per agent, cost optimization opportunities.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/processed/metrics.csv')
df['total_tokens'] = df['tokens_in'] + df['tokens_out']

# Gemini Flash pricing (approximate, $/1M tokens)
PRICE_IN_PER_M = 0.075
PRICE_OUT_PER_M = 0.30

df['cost_usd'] = (df['tokens_in'] / 1e6 * PRICE_IN_PER_M) + (df['tokens_out'] / 1e6 * PRICE_OUT_PER_M)

print('Total cost (USD):', df['cost_usd'].sum().round(4))
print('Avg cost per run (USD):', df['cost_usd'].mean().round(6))

df.groupby('page_type')['total_tokens'].mean().plot(kind='bar', title='Avg Tokens by Page Type')
plt.tight_layout()
plt.show()
```
